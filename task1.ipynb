```python

import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

```

```python
X, y_true = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=1.8,
    random_state=42
)



plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    s=50,
    color='blue'
)

plt.title("Original Artificial Data Points (More Spread Out)")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.grid(True)
plt.show()
```

```python

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

kmeans.fit(X)


# 5. Get Cluster Labels and Centroids
labels = kmeans.labels_
centroids = kmeans.cluster_centers_


# 6. Visualize Clustering Results
plt.figure(figsize=(8, 6))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels,
    cmap='viridis',
    s=50
)

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    color='red',
    marker='X',
    s=250,
    label='Centroids'
)

plt.title("K-Means Clustering Results")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.legend()
plt.grid(True)
plt.show()
```

```python
print("Number of Clusters:", kmeans.n_clusters)

print("\nCentroids:")
print(centroids)

print("\nCluster Labels:")
print(labels)

print("\nInertia:", kmeans.inertia_)

print("\nSilhouette Score:")
from sklearn.metrics import silhouette_score

print(silhouette_score(X, labels))
```
